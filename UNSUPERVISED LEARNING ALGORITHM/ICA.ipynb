```python
from sklearn.decomposition import FastICA
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Data sources link = https://github.com/vincentarelbundock/Rdatasets/blob/master/csv/datasets/attitude.csv

data = pd.read_csv('attitude.csv')
data.head()
```

```python
data.info()
```

```python
ICA = FastICA(n_components=2)
IndependentComp_values = ICA.fit_transform(data.values)
```

```python
print('##final Independent Components##')
reducedata = pd.DataFrame(data=IndependentComp_values, columns=['IC1', 'IC2'])
print(reducedata.head(10))
```
